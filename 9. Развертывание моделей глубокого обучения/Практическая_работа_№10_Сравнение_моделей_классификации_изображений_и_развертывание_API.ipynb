### **Раздел 1. Подготовка и анализ моделей**


1. Загрузить выбранный датасет для классификации изображений
2. Загрузить ранее обученные модели (из практических работ 2-5)
3. Сравнить модели по метрикам качества:
   - Точность (accuracy)
   - Полнота (recall)
   - Precision
   - F1-мера
   - Время инференса (Время, затрачиваемое моделью на формирование предсказания. Измеряется путём усреднения результатов на нескольких тысячах изображений)
4. Создать сводный DataFrame с результатами всех моделей
5. Визуализировать результаты сравнения с помощью графиков
6. Построить матрицы ошибок для каждой модели
7. Выбрать лучшую модель по F1-мере

````python
import os
import time
import random
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from pathlib import Path

import tensorflow as tf
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, Activation, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.optimizers import Adam, SGD
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.preprocessing.image import load_img, img_to_array

from sklearn.preprocessing import LabelBinarizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

import warnings
warnings.filterwarnings('ignore')

print("Библиотеки успешно импортированы.")

DATASET_PATH = '/content/drive/MyDrive/animals/train'
VAL_PATH = '/content/drive/MyDrive/animals/val'

# Параметры изображений
IMG_SIZE = 64
INPUT_SHAPE = (IMG_SIZE, IMG_SIZE, 3) # Для CNN
INPUT_DIM = IMG_SIZE * IMG_SIZE * 3  # Для Dense

# Параметры обучения
BATCH_SIZE = 64
EPOCHS = 50      # Количество эпох можно увеличить для лучшего качества
LEARNING_RATE = 0.0001

# Пути для сохранения моделей
MODEL_SAVE_PATH = '/content/models_comparison'
DENSE_MODEL_PATH = os.path.join(MODEL_SAVE_PATH, 'model_dense.h5')
CNN_SIMPLE_MODEL_PATH = os.path.join(MODEL_SAVE_PATH, 'model_cnn_simple.h5')
CNN_VGG_MODEL_PATH = os.path.join(MODEL_SAVE_PATH, 'model_cnn_vgg.h5')
BEST_MODEL_FILENAME = 'best_classification_model.h5'

# Создаем директорию для моделей
os.makedirs(MODEL_SAVE_PATH, exist_ok=True)

# --- 3. Функции для загрузки и подготовки данных ---
def load_dataset_cnn(base_path, img_size=64):
    """
    Загружает изображения из папок в формате (N, H, W, C) для CNN.
    """
    data = []
    labels = []

    classes = sorted([d for d in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, d))])

    print(f"📁 Найдены классы в {base_path}: {classes}")

    for class_name in classes:
        class_path = os.path.join(base_path, class_name)
        image_files = [f for f in os.listdir(class_path) if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp'))]

        for img_file in tqdm(image_files, desc=f"  Загрузка '{class_name}'"):
            try:
                img_path = os.path.join(class_path, img_file)
                img = load_img(img_path, target_size=(img_size, img_size))
                img_array = img_to_array(img) / 255.0
                data.append(img_array)
                labels.append(class_name)
            except Exception as e:
                print(f"⚠️ Ошибка загрузки {img_file}: {e}")
                continue
    return np.array(data, dtype='float32'), np.array(labels), classes

# Загружаем данные
print("\n🔄 Загрузка обучающей выборки...")
X_train, y_train_str, class_names = load_dataset_cnn(DATASET_PATH, IMG_SIZE)

print("\n🔄 Загрузка валидационной выборки...")
X_val, y_val_str, _ = load_dataset_cnn(VAL_PATH, IMG_SIZE)

print(f"\n✅ Данные загружены: Train: {X_train.shape}, Val: {X_val.shape}")

# Кодируем метки
lb = LabelBinarizer()
y_train = lb.fit_transform(y_train_str)
y_val = lb.transform(y_val_str)
num_classes = len(lb.classes_)
print(f"Метки закодированы. Количество классов: {num_classes}")

# Для Dense модели нужно подготовить плоский вектор
X_train_dense = X_train.reshape(X_train.shape[0], INPUT_DIM)
X_val_dense = X_val.reshape(X_val.shape[0], INPUT_DIM)

# --- 4. Создание и обучение моделей ---

class ModelTrainer:
    """Класс для обучения и сохранения моделей."""
    def __init__(self, name, save_path, input_shape, input_dense_shape, num_classes, epochs=EPOCHS, batch_size=BATCH_SIZE, lr=LEARNING_RATE):
        self.name = name
        self.save_path = save_path
        self.input_shape = input_shape
        self.input_dense_shape = input_dense_shape
        self.num_classes = num_classes
        self.epochs = epochs
        self.batch_size = batch_size
        self.lr = lr
        self.model = None

    def create_model(self):
        raise NotImplementedError("Метод create_model должен быть переопределен в дочернем классе.")

    def get_callbacks(self):
        return [
            EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True, verbose=1),
            ModelCheckpoint(self.save_path, monitor='val_accuracy', save_best_only=True, verbose=1),
            ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=4, min_lr=1e-7, verbose=1)
        ]

    def train(self, X_train, y_train, X_val, y_val):
        if self.model is None:
            self.create_model()
        self.model.compile(optimizer=Adam(learning_rate=self.lr),
                           loss='categorical_crossentropy',
                           metrics=['accuracy'])
        self.model.summary()
        history = self.model.fit(X_train, y_train, validation_data=(X_val, y_val),
                                 epochs=self.epochs, batch_size=self.batch_size,
                                 callbacks=self.get_callbacks(), verbose=1)
        self.model.load_weights(self.save_path) # Загружаем лучшие веса
        return history

class DenseModelTrainer(ModelTrainer):
    """Тренер для полносвязной модели."""
    def create_model(self):
        self.model = Sequential([
            Dense(1024, input_dim=self.input_dense_shape, activation='relu'),
            Dense(512, activation='relu'),
            Dense(256, activation='relu'),
            Dense(self.num_classes, activation='softmax')
        ])

class BaseCNNTrainer(ModelTrainer):
    """Базовый класс для CNN."""
    def get_callbacks(self):
        callbacks = super().get_callbacks()
        return callbacks

    def train(self, X_train, y_train, X_val, y_val):
        # CNN принимают данные в формате (N, H, W, C)
        return super().train(X_train, y_train, X_val, y_val)

class SimpleCNNTrainer(BaseCNNTrainer):
    """Тренер для простой CNN."""
    def create_model(self):
        self.model = Sequential([
            Conv2D(32, (3,3), activation='relu', padding='same', input_shape=self.input_shape),
            MaxPooling2D((2,2)),
            Conv2D(64, (3,3), activation='relu', padding='same'),
            MaxPooling2D((2,2)),
            Conv2D(128, (3,3), activation='relu', padding='same'),
            MaxPooling2D((2,2)),
            Flatten(),
            Dense(256, activation='relu'),
            Dense(self.num_classes, activation='softmax')
        ])

class VGGCNNTrainer(BaseCNNTrainer):
    """Тренер для VGG-подобной CNN."""
    def create_model(self):
        self.model = Sequential([
            # Блок 1
            Conv2D(64, (3,3), activation='relu', padding='same', input_shape=self.input_shape),
            Conv2D(64, (3,3), activation='relu', padding='same'),
            MaxPooling2D((2,2)),
            # Блок 2
            Conv2D(128, (3,3), activation='relu', padding='same'),
            Conv2D(128, (3,3), activation='relu', padding='same'),
            MaxPooling2D((2,2)),
            # Блок 3
            Conv2D(256, (3,3), activation='relu', padding='same'),
            Conv2D(256, (3,3), activation='relu', padding='same'),
            MaxPooling2D((2,2)),
            # Блок 4
            Conv2D(512, (3,3), activation='relu', padding='same'),
            Conv2D(512, (3,3), activation='relu', padding='same'),
            MaxPooling2D((2,2)),
            # Классификатор
            Flatten(),
            Dense(512, activation='relu'),
            Dense(self.num_classes, activation='softmax')
        ])

# Запуск обучения для всех моделей
print("\n\n" + "="*20 + "🧠 ОБУЧЕНИЕ DENSE МОДЕЛИ " + "="*20)
dense_trainer = DenseModelTrainer('Dense', DENSE_MODEL_PATH, INPUT_SHAPE, INPUT_DIM, num_classes)
history_dense = dense_trainer.train(X_train_dense, y_train, X_val_dense, y_val)

print("\n\n" + "="*20 + "🧠 ОБУЧЕНИЕ SIMPLE CNN " + "="*20)
simple_cnn_trainer = SimpleCNNTrainer('SimpleCNN', CNN_SIMPLE_MODEL_PATH, INPUT_SHAPE, None, num_classes)
history_cnn = simple_cnn_trainer.train(X_train, y_train, X_val, y_val)

print("\n\n" + "="*20 + "🧠 ОБУЧЕНИЕ VGG CNN " + "="*20)
vgg_cnn_trainer = VGGCNNTrainer('VGGCNN', CNN_VGG_MODEL_PATH, INPUT_SHAPE, None, num_classes)
history_vgg = vgg_cnn_trainer.train(X_train, y_train, X_val, y_val)

# --- 5. Функции для оценки и сравнения моделей ---
def evaluate_model(model, X_val, y_val, X_train_sample, class_names, model_type='cnn'):
    """Оценивает модель по ключевым метрикам и возвращает словарь результатов."""
    start_time = time.time()
    if model_type == 'dense':
        y_pred_probs = model.predict(X_val, verbose=0)
    else:
        y_pred_probs = model.predict(X_val.reshape(-1, *INPUT_SHAPE), verbose=0)
    inference_time = time.time() - start_time

    y_pred = np.argmax(y_pred_probs, axis=1)
    y_true = np.argmax(y_val, axis=1)

    report = classification_report(y_true, y_pred, target_names=class_names, output_dict=True)

    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, average='weighted'),
        'Recall': recall_score(y_true, y_pred, average='weighted'),
        'F1-Score': f1_score(y_true, y_pred, average='weighted'),
        'Inference Time (s)': inference_time / len(X_val), # Время на одно изображение
        'Report': report,
        'y_pred': y_pred,
        'y_true': y_true
    }

# Оценка всех моделей
print("\n\n🚀 Оценка моделей...")
results = {}
results['Dense'] = evaluate_model(dense_trainer.model, X_val_dense, y_val, X_val, class_names, model_type='dense')
results['SimpleCNN'] = evaluate_model(simple_cnn_trainer.model, X_val, y_val, X_val, class_names)
results['VGGCNN'] = evaluate_model(vgg_cnn_trainer.model, X_val, y_val, X_val, class_names)

# --- 6. Создание сводного DataFrame ---
df_results = pd.DataFrame.from_dict({
    model: {k: v for k, v in metrics.items() if k not in ['Report', 'y_pred', 'y_true']}
    for model, metrics in results.items()
}, orient='index')

df_results = df_results[['Accuracy', 'Precision', 'Recall', 'F1-Score', 'Inference Time (s)']]
df_results['Inference Time (ms)'] = df_results['Inference Time (s)'] * 1000

print("\n📊 Сводная таблица результатов:")
print(df_results)

# --- 7. Визуализация результатов ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# График метрик
metrics_to_plot = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
df_metrics = df_results[metrics_to_plot]
df_metrics.plot(kind='bar', ax=axes[0], colormap='viridis')
axes[0].set_title('Сравнение метрик качества')
axes[0].set_ylabel('Score')
axes[0].set_ylim([0, 1])
axes[0].tick_params(axis='x', rotation=0)
axes[0].legend(loc='lower right')

# График времени инференса
df_results['Inference Time (ms)'].plot(kind='bar', ax=axes[1], color='orange')
axes[1].set_title('Среднее время инференса на изображение')
axes[1].set_ylabel('Время (ms)')
axes[1].tick_params(axis='x', rotation=0)

# График точности во времени обучения
trainer_histories = {
    'Dense': dense_trainer,
    'SimpleCNN': simple_cnn_trainer,
    'VGGCNN': vgg_cnn_trainer
}
for name, trainer in trainer_histories.items():
    axes[2].plot(trainer.history.history['val_accuracy'], label=name)
axes[2].set_title('Динамика точности на валидации при обучении')
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('Val Accuracy')
axes[2].legend()
axes[2].grid(True)

plt.tight_layout()
plt.show()

# --- 8. Матрицы ошибок ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, (name, result) in enumerate(results.items()):
    cm = confusion_matrix(result['y_true'], result['y_pred'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[i])
    axes[i].set_title(f'Матрица ошибок - {name}')
    axes[i].set_xlabel('Predicted Label')
    axes[i].set_ylabel('True Label')
plt.tight_layout()
plt.show()

# --- 9. Выбор и сохранение лучшей модели ---
best_model_name = df_results['F1-Score'].idxmax()
print(f"\n🏆 Лучшая модель по F1-мере: {best_model_name} (F1 = {df_results.loc[best_model_name, 'F1-Score']:.4f})")

# Определяем путь к лучшей модели
trainer_map = {
    'Dense': dense_trainer,
    'SimpleCNN': simple_cnn_trainer,
    'VGGCNN': vgg_cnn_trainer
}
best_trainer = trainer_map[best_model_name]

best_model_destination = os.path.join(MODEL_SAVE_PATH, BEST_MODEL_FILENAME)

if os.path.exists(best_model_destination):
    os.remove(best_model_destination)

shutil.copy(best_trainer.save_path, best_model_destination)

# Сохраняем конфигурацию
import pickle
config = {
    'class_names': class_names,
    'img_size': IMG_SIZE,
    'num_classes': num_classes,
    'input_shape': INPUT_SHAPE,
    'model_type': best_trainer.name,
    'label_binarizer': lb
}
with open(os.path.join(MODEL_SAVE_PATH, 'model_config.pkl'), 'wb') as f:
    pickle.dump(config, f)

print(f"✅ Лучшая модель сохранена в: {best_model_destination}")
````

````python
dense_trainer.history = history_dense
simple_cnn_trainer.history = history_cnn
vgg_cnn_trainer.history = history_vgg

print("✅ Истории обучения присвоены.")

# --- 7. Визуализация результатов ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# График метрик
metrics_to_plot = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
df_metrics = df_results[metrics_to_plot]
df_metrics.plot(kind='bar', ax=axes[0], colormap='viridis')
axes[0].set_title('Сравнение метрик качества')
axes[0].set_ylabel('Score')
axes[0].set_ylim([0, 1])
axes[0].tick_params(axis='x', rotation=0)
axes[0].legend(loc='lower right')

# График времени инференса
df_results['Inference Time (ms)'].plot(kind='bar', ax=axes[1], color='orange')
axes[1].set_title('Среднее время инференса на изображение')
axes[1].set_ylabel('Время (ms)')
axes[1].tick_params(axis='x', rotation=0)

# График точности во времени обучения
trainer_histories = {
    'Dense': dense_trainer,
    'SimpleCNN': simple_cnn_trainer,
    'VGGCNN': vgg_cnn_trainer
}
for name, trainer in trainer_histories.items():
    axes[2].plot(trainer.history.history['val_accuracy'], label=name)
axes[2].set_title('Динамика точности на валидации при обучении')
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('Val Accuracy')
axes[2].legend()
axes[2].grid(True)

plt.tight_layout()
plt.show()

# --- 8. Матрицы ошибок ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, (name, result) in enumerate(results.items()):
    cm = confusion_matrix(result['y_true'], result['y_pred'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[i])
    axes[i].set_title(f'Матрица ошибок - {name}')
    axes[i].set_xlabel('Predicted Label')
    axes[i].set_ylabel('True Label')
plt.tight_layout()
plt.show()

# --- 9. Выбор и сохранение лучшей модели ---
best_model_name = df_results['F1-Score'].idxmax()
print(f"\n🏆 Лучшая модель по F1-мере: {best_model_name} (F1 = {df_results.loc[best_model_name, 'F1-Score']:.4f})")

# Определяем путь к лучшей модели
trainer_map = {
    'Dense': dense_trainer,
    'SimpleCNN': simple_cnn_trainer,
    'VGGCNN': vgg_cnn_trainer
}
best_trainer = trainer_map[best_model_name]

best_model_destination = os.path.join(MODEL_SAVE_PATH, BEST_MODEL_FILENAME)

if os.path.exists(best_model_destination):
    os.remove(best_model_destination)

shutil.copy(best_trainer.save_path, best_model_destination)

# Сохраняем конфигурацию
import pickle
config = {
    'class_names': class_names,
    'img_size': IMG_SIZE,
    'num_classes': num_classes,
    'input_shape': INPUT_SHAPE,
    'model_type': best_trainer.name,
    'label_binarizer': lb
}
with open(os.path.join(MODEL_SAVE_PATH, 'model_config.pkl'), 'wb') as f:
    pickle.dump(config, f)

print(f"✅ Лучшая модель сохранена в: {best_model_destination}")
````

#### **Инструкция по развёртыванию Hugging Face Spaces (если возникли пролемы с доступом к Render.com)**


**Шаги:**

1. **Регистрация:**
   - Зарегистрируйтесь на [huggingface.co](https://huggingface.co) (бесплатно, без карты).
   - Подтвердите e-mail.

2. **Создание нового Space:**
   - На главной странице нажмите на аватар → **New Space**.
   - Заполните поля:
     - **Space name:** `имя_проекта_backend`.
     - **License:** `mit` (или любая на ваш выбор).
     - **SDK:** выбрать **Docker** → **Blank**.
     - **Visibility:** `Public`.
     - **Hardware:** `CPU basic · 2 vCPU · 16 GB · FREE`.

3. **Push репозитория в Hugging Face:**
   - После создания Space Hugging Face даст адрес Git-репозитория вида `https://huggingface.co/spaces/ВАШ_USERNAME/california-regression`.
   - Добавьте его как удалённый репозиторий в ваш локальный проект:
   ```bash
   git remote add hf https://huggingface.co/spaces/ВАШ_USERNAME/ВАШ_РЕПОЗИТОРИЙ
   git push hf main
   ```
   - При push потребуется логин Hugging Face и **Access Token** (создаётся в Settings → Access Tokens с правом `write`).

4. **Автоматическая сборка:**
   - Hugging Face автоматически обнаружит `Dockerfile` и запустит сборку.
   - Процесс сборки занимает **5–10 минут** (установка TensorFlow — самая долгая часть).
   - Статус можно отслеживать на вкладке **Logs** в интерфейсе Space.

5. **Получение публичного URL:**
   - После успешного деплоя API доступен по адресу:
     ```
     https://ВАШ_USERNAME-ИМЯ-ПРОЕКТА.hf.space/predict
     ```
   - Этот URL работает 24/7 без усыпления, в отличие от бесплатных тарифов Render.com.
